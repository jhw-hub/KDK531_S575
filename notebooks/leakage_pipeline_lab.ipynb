# Leakage-safe preprocessing exercise
Original synthetic data. The first approach is deliberately wrong; **do not use it for model assessment**. Identify why scaling on all observations leaks into validation.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
X,y=make_classification(n_samples=240,n_features=5,n_informative=3,random_state=531)
cv=StratifiedKFold(n_splits=4,shuffle=True,random_state=575)
# WRONG: fit scaler before CV (validation fold leaks into scaling statistics)
wrong=StandardScaler().fit_transform(X)
print("Leaky score, do not use:",cross_val_score(LogisticRegression(max_iter=500),wrong,y,cv=cv,scoring="roc_auc").mean())
# RIGHT: fit preprocessing within training folds
pipe=make_pipeline(StandardScaler(),LogisticRegression(max_iter=500))
print("Fold-safe score:",cross_val_score(pipe,X,y,cv=cv,scoring="roc_auc").mean())
```
