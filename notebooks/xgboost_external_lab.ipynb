```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, confusion_matrix
from xgboost import XGBClassifier
X, y = make_classification(n_samples=600, n_features=6, n_informative=4, n_redundant=1, weights=[0.85,0.15], flip_y=0.06, random_state=531575)
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,stratify=y,random_state=575)
model = XGBClassifier(n_estimators=50,max_depth=2,learning_rate=0.06,random_state=575,n_jobs=1,eval_metric="logloss")
model.fit(X_train,y_train)
prob=model.predict_proba(X_test)[:,list(model.classes_).index(1)]
print("Positive class:", 1, "class order:", model.classes_)
print("Holdout AUC:",round(roc_auc_score(y_test,prob),3))
print("Confusion matrix rows true 0/1, columns pred 0/1:\n",confusion_matrix(y_test,prob>=0.3,labels=[0,1]))
```
